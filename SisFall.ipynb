```python
def extract_from_high_amp_segments(
    data_list, window_size, step_size, amp_scale, min_gap, before, after, sensor_type='acc'
):
    """
    For each signal in data_list, extract up to two segments where the peak-to-peak amplitude
    is above a threshold (relative to max). Segments are separated by at least `min_gap`.

    Parameters:
        data_list: list of np.ndarray, each of shape (6, n_samples)
        window_size: int, size of each segment to extract
        step_size: int, step size for sliding window
        amp_scale: float, threshold as a fraction of max peak-to-peak value
        min_gap: int, minimum number of samples between two selected segments
        before: int, number of samples before peak to include
        after: int, total number of samples in the segment
        sensor_type: str, 'acc' or 'gyro' to select which sensor to use for peak detection

    Returns:
        list_1: list of first high-amplitude segments
        list_2: list of second high-amplitude segments (or empty if only one segment found)
    """
    list_1 = []
    list_2 = []

    # Choose axes based on sensor_type
    if sensor_type == 'acc':
        selected_data = [signal[0:3, :] for signal in data_list]  # Rows 0–2: Accelerometer
    elif sensor_type == 'gyro':
        selected_data = [signal[3:6, :] for signal in data_list]  # Rows 3–5: Gyroscope
    else:
        raise ValueError("sensor_type must be 'acc' or 'gyro'")

    # Compute peak-to-peak amplitude features
    amp_features = fn.max_peak_to_peak_amp(selected_data, window_size, step_size)

    for signal_idx, signal in enumerate(data_list):
        features = amp_features[signal_idx]
        max_val = max(features)
        threshold = amp_scale * max_val

        selected_segments = []
        last_added_index = -min_gap  # so first one can always be added

        for i, value in enumerate(features):
            if value >= threshold:
                segment_start = i * step_size - before
                segment_end = segment_start + after

                if segment_start < 0 :
                    continue  # skip invalid windows
                if segment_end > signal.shape[1]:
                    segment_end = signal.shape[1]

                # Avoid overlap by checking distance to last segment
                if segment_start - last_added_index >= min_gap:
                    selected_segments.append(signal[:, segment_start:segment_end])
                    last_added_index = segment_start

            if len(selected_segments) == 2:
                break

        # Append segments or empty arrays if not enough were found
        if len(selected_segments) >= 1:
            list_1.append(selected_segments[0])
        else:
            list_1.append(np.empty((6, 0)))

        if len(selected_segments) >= 2:
            list_2.append(selected_segments[1])
        else:
            list_2.append(np.empty((6, 0)))

    lengths1 = [s.shape[1] for s in list_1]
    lengths2 = [s.shape[1] for s in list_2]
    print(f"{len(list_1)} files processed using scale: {amp_scale}"
          f" \nMin1: {min(lengths1)}, Max1: {max(lengths1)}"
          f" \nMin2: {min(lengths2)}, Max2: {max(lengths2)}")

    return list_1, list_2
```

```python
def keep_high_amp_segments_separated2(
    data_list, window_size, step_size, amp_scale, min_gap, before, after, sensor_type='acc'
):
    """
    For each signal in data_list, extract up to two segments where the peak-to-peak amplitude
    is above a threshold (relative to max). Segments are separated by at least `min_gap`.

    Parameters:
        data_list: list of np.ndarray, each of shape (6, n_samples)
        window_size: int, size of each segment to extract
        step_size: int, step size for sliding window
        amp_scale: float, threshold as a fraction of max peak-to-peak value
        min_gap: int, minimum number of samples between two selected segments
        before: int, number of samples before peak to include
        after: int, total number of samples in the segment
        sensor_type: str, 'acc' or 'gyro' to select which sensor to use for peak detection

    Returns:
        list_1: list of first high-amplitude segments
        list_2: list of second high-amplitude segments (or empty if only one segment found)
    """
    list_1 = []
    list_2 = []

    # Choose axes based on sensor_type
    if sensor_type == 'acc':
        selected_data = [signal[0:3, :] for signal in data_list]  # Rows 0–2: Accelerometer
    elif sensor_type == 'gyro':
        selected_data = [signal[3:6, :] for signal in data_list]  # Rows 3–5: Gyroscope
    else:
        raise ValueError("sensor_type must be 'acc' or 'gyro'")

    # Compute peak-to-peak amplitude features
    amp_features = fn.max_peak_to_peak_amp(selected_data, window_size, step_size)

    for signal_idx, signal in enumerate(data_list):
        features = amp_features[signal_idx]
        max_val = max(features)
        threshold = amp_scale * max_val

        selected_segments = []
        last_added_index = -min_gap  # so first one can always be added

        for i, value in enumerate(features):
            if value >= threshold:
                segment_start = i * step_size - before
                segment_end = segment_start + after

                if segment_start < 0 :
                    continue  # skip invalid windows
                if segment_end > signal.shape[1]:
                    segment_end = signal.shape[1]

                # Avoid overlap by checking distance to last segment
                if segment_start - last_added_index >= min_gap:
                    selected_segments.append(signal[:, segment_start:segment_end])
                    last_added_index = segment_start

            if len(selected_segments) == 2:
                break

        # Append segments or empty arrays if not enough were found
        if len(selected_segments) >= 1:
            list_1.append(selected_segments[0])
        else:
            list_1.append(np.empty((6, 0)))

        if len(selected_segments) >= 2:
            list_2.append(selected_segments[1])
        else:
            list_2.append(np.empty((6, 0)))

    lengths1 = [s.shape[1] for s in list_1]
    lengths2 = [s.shape[1] for s in list_2]
    print(f"{len(list_1)} files processed using scale: {amp_scale}"
          f" \nMin1: {min(lengths1)}, Max1: {max(lengths1)}"
          f" \nMin2: {min(lengths2)}, Max2: {max(lengths2)}")

    return list_1, list_2
```

```python
def keep_high_amp_segments_separated(data_list, window_size, step_size, amp_scale, min_gap, before, after):
    """
    For each signal in data_list, extract up to two segments where the peak-to-peak amplitude
    is above a threshold (relative to max). Segments are separated by at least `min_gap`.

    Parameters:
        data_list: list of np.ndarray, each of shape (6, n_samples)
        window_size: int, size of each segment to extract
        step_size: int, step size for sliding window
        amp_scale: float, threshold as a fraction of max peak-to-peak value
        min_gap: int, minimum number of samples between two selected segments

    Returns:
        list_1: list of first high-amplitude segments
        list_2: list of second high-amplitude segments (or empty if only one segment found)
    """
    list_1 = []
    list_2 = []

    # Compute peak-to-peak amplitude features
    amp_features = fn.max_peak_to_peak_amp(data_list, window_size, step_size)

    for signal_idx, signal in enumerate(data_list):
        features = amp_features[signal_idx]
        max_val = max(features)
        threshold = amp_scale * max_val

        selected_segments = []
        last_added_index = -min_gap  # so first one can always be added

        for i, value in enumerate(features):
            if value >= threshold:
                segment_start = i * step_size - before
                segment_end = segment_start + after

                # Avoid overlap by checking distance to last segment
                if segment_start - last_added_index >= min_gap:
                    # Clip to signal length bounds
                    if segment_end <= signal.shape[1]:
                        selected_segments.append(signal[:, segment_start:segment_end])
                        last_added_index = segment_start

            if len(selected_segments) == 2:
                break

        # Append segments or empty arrays if not enough were found
        if len(selected_segments) >= 1:
            list_1.append(selected_segments[0])
        else:
            list_1.append(np.empty((6, 0)))

        if len(selected_segments) >= 2:
            list_2.append(selected_segments[1])
        else:
            list_2.append(np.empty((6, 0)))

    lengths1 = [s.shape[1] for s in list_1]
    lengths2 = [s.shape[1] for s in list_2]
    print(f"{len(list_1)} files processed using scale: {amp_scale}"
          f" \nMin1: {min(lengths1)}, Max1: {max(lengths1)}"
          f" \nMin2: {min(lengths2)}, Max2: {max(lengths2)}")
    
    # print(f"Processed {len(data_list)} signals. Extracted up to 2 segments each.")
    return list_1, list_2
```

```python
def keep_from_peak_2(data_list, window_size, peak_threshold_scale=0.8, min_gap=None):
    """
    For each signal in data_list, extract up to two movement segments centered around peaks in the
    combined absolute value of the first 3 axes. Peaks must be above a threshold scale of the global max.

    Returns:
        list_1: list of first detected peak segments
        list_2: list of second detected peak segments
    """
    if min_gap is None:
        min_gap = window_size  # Default to window_size if not given

    list_1 = []
    list_2 = []

    for signal in data_list:
        combined = np.abs(signal[0]) + np.abs(signal[1]) + np.abs(signal[2])
        peak_value = np.max(combined)
        threshold = peak_value * peak_threshold_scale

        # Find all indices where the combined signal exceeds the threshold
        above_th_indices = np.where(combined > threshold)[0]

        # Group continuous indices into separate peaks
        peak_indices = []
        if len(above_th_indices) > 0:
            current_peak = [above_th_indices[0]]
            for idx in above_th_indices[1:]:
                if idx - current_peak[-1] <= min_gap:
                    current_peak.append(idx)
                else:
                    peak_indices.append(current_peak)
                    current_peak = [idx]
            peak_indices.append(current_peak)

        # Extract up to two windows centered around the main points in each peak group
        peak_segments = []
        for peak_group in peak_indices[:2]:  # Only take up to 2 peaks
            center = peak_group[len(peak_group) // 2]
            start = max(0, center - window_size // 2)
            end = min(signal.shape[1], start + window_size)
            segment = signal[:, start:end]

            # Pad if needed to maintain window size
            if segment.shape[1] < window_size:
                pad_width = window_size - segment.shape[1]
                segment = np.pad(segment, ((0,0),(0,pad_width)), mode='constant')

            peak_segments.append(segment)

        # Store the two segments separately
        list_1.append(peak_segments[0] if len(peak_segments) > 0 else np.empty((6, 0)))
        list_2.append(peak_segments[1] if len(peak_segments) > 1 else np.empty((6, 0)))

    # Print stats
    min1 = min(seg.shape[1] for seg in list_1)
    max1 = max(seg.shape[1] for seg in list_1)
    min2 = min(seg.shape[1] for seg in list_2)
    max2 = max(seg.shape[1] for seg in list_2)
    print(f"{len(list_1)} files processed.")
    print(f"Part 1: Min length = {min1}, Max length = {max1}")
    print(f"Part 2: Min length = {min2}, Max length = {max2}")

    return list_1, list_2
```

```python
def keep_high_std_segments_separated(data_list, window_size, step_size, peak_threshold_scale, before, after):
    """
    For each signal in data_list, extract up to two high-STD segments separately.

    Parameters:
        data_list: list of np.ndarray, each of shape (6, n_samples)
        window_size: int - sliding window size
        step_size: int - stride for the sliding window
        peak_threshold_scale: float - minimum STD magnitude to consider as a peak (relative to max)

    Returns:
        list_1: list of np.ndarray (first detected segments)
        list_2: list of np.ndarray (second detected segments)
    """
    list_1 = []
    list_2 = []

    for signal in data_list:
        acc_x = signal[0, :]
        acc_z = signal[2, :]
        n = acc_x.shape[0]

        values = []
        indices = []

        # Compute STD magnitude over sliding windows
        for i in range(0, n - window_size + 1, step_size):
            segment_x = acc_x[i:i + window_size]
            segment_z = acc_z[i:i + window_size]

            std_x = np.std(segment_x)
            std_z = np.std(segment_z)

            value = np.sqrt(std_x**2 + std_z**2)
            values.append(value)
            indices.append(i)

        values = np.array(values)
        indices = np.array(indices)

        # Find peaks above threshold
        max_val = np.max(values)
        threshold = max_val * peak_threshold_scale
        high_indices = indices[values >= threshold]

        # Remove peaks that are too close to each other (non-overlapping windows)
        kept_segments = []
        used = np.zeros_like(signal[0], dtype=bool)

        for idx in high_indices:
            start = idx - before
            end = idx + after

            # Skip if overlapping with already selected region
            if np.any(used[start:end]):
                continue

            segment = signal[:, start:end]
            kept_segments.append(segment)
            used[start:end] = True

            if len(kept_segments) == 2:
                break

        # Save to two separate lists
        if len(kept_segments) > 0:
            list_1.append(kept_segments[0])
        else:
            list_1.append(np.empty((6, 0)))

        if len(kept_segments) > 1:
            list_2.append(kept_segments[1])
        else:
            list_2.append(np.empty((6, 0)))
    # Print min/max lengths
    lengths1 = [s.shape[1] for s in list_1]
    lengths2 = [s.shape[1] for s in list_2]
    print(f"{len(list_1)} files processed using {peak_threshold_scale} scale"
          f" \nMin1: {min(lengths1)}, Max1: {max(lengths1)}"
          f" \nMin2: {min(lengths2)}, Max2: {max(lengths2)}")
    return list_1, list_2
```

```python
# calculate fft of signal
def calculate_fft_2(signal, fs):
    N = len(signal)
    T = 1.0 / fs  # Sampling interval (assuming 50 Hz sampling rate)
    yf = np.fft.fft(signal)
    xf = np.fft.fftfreq(N, T)[:N // 2]
    return xf, 2.0 / N * np.abs(yf[0:N // 2])
```

```python
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.utils import to_categorical
```

```python
# Split into Train/Test Sets for Convolutional Neural Network model

# Split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(trimmed_data, boolean_list, test_size=0.4, random_state=42)
# Split the testing set into validation and testing sets
x_val, x_test, y_val, y_test = train_test_split(x_test, y_test, test_size=0.5, random_state=42)
print(f"Training data shape: {x_train.shape}")
print(f"Testing data shape: {x_test.shape}")
print(f"Validation data shape: {x_val.shape}")

# Define the number of classes
num_classes = 2  # Binary classification: Fall (0) and ADL (1)

# Reshape data for CNN input shape
x_train = x_train.reshape(-1, 6, min_length, 1)
x_val = x_val.reshape(-1, 6, min_length, 1)
x_test = x_test.reshape(-1, 6, min_length, 1)

# Convert labels to one-hot encoding
from keras.utils import to_categorical
y_train = to_categorical(y_train, num_classes=num_classes)
y_val = to_categorical(y_val, num_classes=num_classes)
y_test = to_categorical(y_test, num_classes=num_classes)

# Build the CNN model
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(6, min_length, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(num_classes, activation='softmax')  # For categorical classification
])

model.summary()

# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Train the model
history = model.fit(x_train, y_train, epochs=10, batch_size=32, validation_data=(x_val, y_val), verbose=1)
```

```python
# evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
```

```python
def extract_from_high_amp_segments(
    data_list, window_size, step_size, amp_scale, min_gap,
    before_1, before_2, after, sensor_type='acc'
):
    """
    Extracts up to two high-amplitude segments per signal, with customizable window positions.

    Parameters:
        data_list: list of np.ndarray, each (6, n_samples)
        window_size: int, window size for peak detection
        step_size: int, step for sliding window
        amp_scale: float, threshold relative to max peak
        min_gap: int, minimum gap between segments
        before_1, after_1: int, samples before and after peak for first segment
        before_2, after_2: int, samples before and after peak for second segment
        sensor_type: 'acc' or 'gyro'

    Returns:
        list_1: list of first high-amplitude segments
        list_2: list of second high-amplitude segments
    """
    list_1 = []
    list_2 = []

    # Choose axes based on sensor type
    if sensor_type == 'acc':
        selected_data = [signal[0:3, :] for signal in data_list]
    elif sensor_type == 'gyro':
        selected_data = [signal[3:6, :] for signal in data_list]
    else:
        raise ValueError("sensor_type must be 'acc' or 'gyro'")

    # Compute feature
    selected_features = fn.standard_deviation_magnitude(selected_data, window_size, step_size)

    for signal_idx, signal in enumerate(data_list):
        features = selected_features[signal_idx]
        max_val = max(features)
        threshold = amp_scale * max_val

        selected_segments = []
        last_added_index = -min_gap

        for i, value in enumerate(features):
            if value >= threshold:
                if len(selected_segments) == 0:
                    segment_start = i * step_size - before_1
                    segment_end = segment_start + after
                elif len(selected_segments) == 1:
                    segment_start = i * step_size - before_2
                    segment_end = segment_start + after
                else:
                    break  # already have two

                if segment_start < 0:
                    continue
                if segment_end > signal.shape[1]:
                    segment_end = signal.shape[1]

                if segment_start - last_added_index >= min_gap:
                    selected_segments.append(signal[:, segment_start:segment_end])
                    last_added_index = segment_start

        if len(selected_segments) >= 1:
            list_1.append(selected_segments[0])
        else:
            list_1.append(np.empty((6, 0)))

        if len(selected_segments) >= 2:
            list_2.append(selected_segments[1])
        else:
            list_2.append(np.empty((6, 0)))

    lengths1 = [s.shape[1] for s in list_1]
    lengths2 = [s.shape[1] for s in list_2]
    print(f"{len(list_1)} files processed using scale: {amp_scale}"
          f"\nMin1: {min(lengths1)}, Max1: {max(lengths1)}"
          f"\nMin2: {min(lengths2)}, Max2: {max(lengths2)}")

    return list_1, list_2
```

```python
def extract_segments_normalized(
    data_list, window_size, step_size, threshold, min_gap,
    before_1, before_2, after, sensor_type='acc'
):
    list_1 = []
    list_2 = []

    # Select relevant data
    selected_data = [signal[0:3, :] if sensor_type == 'acc' else signal[3:6, :] for signal in data_list]

    # Extract features
    selected_features = fn.standard_deviation_magnitude(selected_data, window_size, step_size)

    for signal_idx, signal in enumerate(data_list):
        features = selected_features[signal_idx]
        # Normalize
        features_norm = (np.array(features) - min(features)) / (max(features) - min(features) + 1e-8)

        selected_segments = []
        last_added_index = -min_gap

        for i, value in enumerate(features_norm):
            if value >= threshold:  # directly use amp_scale as threshold
                if len(selected_segments) == 0:
                    segment_start = i * step_size - before_1
                    segment_end = segment_start + after
                elif len(selected_segments) == 1:
                    segment_start = i * step_size - before_2
                    segment_end = segment_start + after
                else:
                    break

                if segment_start < 0 or segment_end > signal.shape[1]:
                    continue

                if segment_start - last_added_index >= min_gap:
                    selected_segments.append(signal[:, segment_start:segment_end])
                    last_added_index = segment_start

        list_1.append(selected_segments[0] if len(selected_segments) > 0 else np.empty((6, 0)))
        list_2.append(selected_segments[1] if len(selected_segments) > 1 else np.empty((6, 0)))
    
    lengths1 = [s.shape[1] for s in list_1]
    lengths2 = [s.shape[1] for s in list_2]
    print(f"{len(list_1)} files processed using scale: {threshold}"
          f" \nMin1: {min(lengths1)}, Max1: {max(lengths1)}"
          f" \nMin2: {min(lengths2)}, Max2: {max(lengths2)}")
    return list_1, list_2
```

```python
def extract_segments_top2_dynamic(
    data_list, window_size, step_size, min_gap,
    before_1, before_2, after, sensor_type='acc'
):
    list_1 = []
    list_2 = []

    # Select relevant data
    selected_data = [signal[0:3, :] if sensor_type == 'acc' else signal[3:6, :] for signal in data_list]

    # Extract features
    selected_features = fn.standard_deviation_magnitude(selected_data, window_size, step_size)

    for signal_idx, signal in enumerate(data_list):
        features = selected_features[signal_idx]
        features_array = np.array(features)

        # Get indices of top-2 values
        top_indices = features_array.argsort()[-2:][::-1]
        top_indices.sort()  # Ensure they are in order for min_gap checking

        selected_segments = []
        last_added_index = -min_gap

        for seg_num, i in enumerate(top_indices):
            segment_start = i * step_size - (before_1 if seg_num == 0 else before_2)
            segment_end = segment_start + after

            if segment_start < 0 or segment_end > signal.shape[1]:
                continue

            if segment_start - last_added_index >= min_gap:
                selected_segments.append(signal[:, segment_start:segment_end])
                last_added_index = segment_start

        list_1.append(selected_segments[0] if len(selected_segments) > 0 else np.empty((6, 0)))
        list_2.append(selected_segments[1] if len(selected_segments) > 1 else np.empty((6, 0)))

    lengths1 = [s.shape[1] for s in list_1]
    lengths2 = [s.shape[1] for s in list_2]
    print(f"{len(list_1)} files processed"
          f" \nMin1: {min(lengths1)}, Max1: {max(lengths1)}"
          f" \nMin2: {min(lengths2)}, Max2: {max(lengths2)}")
    return list_1, list_2
```

```python
from scipy.signal import find_peaks

def extract_segments_normalized_peaks(
    data_list, window_size, step_size, threshold, min_gap,
    before_1, before_2, after, sensor_type
):
    list_1 = []
    list_2 = []

    # Select relevant data
    selected_data = [signal[0:3, :] if sensor_type == 'acc' else signal[3:6, :] for signal in data_list]

    # Extract features
    selected_features = fn.standard_deviation_magnitude(selected_data, window_size, step_size)

    for signal_idx, signal in enumerate(data_list):
        features = selected_features[signal_idx]
        features = np.array(features)

        # Normalize
        features_norm = (features - np.min(features)) / (np.max(features) - np.min(features) + 1e-8)

        # Find peaks above threshold and with min_gap distance (in samples)
        peaks, _ = find_peaks(features_norm, height=threshold, distance=min_gap // step_size)

        # Sort peaks by height and take top 2
        if len(peaks) > 0:
            sorted_peaks = peaks[np.argsort(features_norm[peaks])][::-1]  # Descending by height
        else:
            sorted_peaks = []

        selected_segments = []
        for i, peak_idx in enumerate(sorted_peaks[:2]):
            segment_start = peak_idx * step_size - (before_1 if i == 0 else before_2)
            segment_end = segment_start + after

            if segment_start < 0 or segment_end > signal.shape[1]:
                continue

            selected_segments.append(signal[:, segment_start:segment_end])

        list_1.append(selected_segments[0] if len(selected_segments) > 0 else np.empty((6, 0)))
        list_2.append(selected_segments[1] if len(selected_segments) > 1 else np.empty((6, 0)))

    lengths1 = [s.shape[1] for s in list_1]
    lengths2 = [s.shape[1] for s in list_2]
    print(f"{len(list_1)} files processed using scale: {threshold}"
          f"\nMin1: {min(lengths1)}, Max1: {max(lengths1)}"
          f"\nMin2: {min(lengths2)}, Max2: {max(lengths2)}")

    return list_1, list_2
```
